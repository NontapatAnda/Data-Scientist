# Data Visualization

```python
df_sorted = df_cleaned.sort_values(by="Global_Sales",ascending=False)
df_top = df_sorted.head(20)
df_top = df_top.set_index("Name")
```

```python
#Best game sales
df_top["Global_Sales"].plot(kind ='barh', figsize=(10,6))
```

```python
# Group by Year and sum the Global_Sales
sales_by_year = df_cleaned.groupby('Year')['Global_Sales'].sum().sort_index()
sales_by_year
```

```python
# Plot Group by Year and sum the
my_title = "Total Global Game Sales by Year"
plt.figure(figsize=(12,6))
plt.plot(sales_by_year.index,sales_by_year.values)
plt.title(my_title)
plt.xlabel('Year')
plt.ylabel('Global Sales (millions)')
plt.show()
```

```python
#Best game sales consider genre
sales_by_genre = df_cleaned.groupby('Genre')['Global_Sales'].sum().sort_index()
sales_by_genre
```

```python
#Plot Best game sales consider genre
plt.figure(figsize=(12,6))
plt.bar(sales_by_genre.index,sales_by_genre.values, color = 'skyblue')
```

```python
#Which Platform do NA consider?
platform_by_Na = df_cleaned.groupby('Platform')['NA_Sales'].sum().sort_values(ascending=False).head(10)
platform_by_Na
```

```python
plt.figure(figsize=(15,8))
plt.scatter(x=platform_by_Na.index,y=platform_by_Na.values,color='darkblue')
plt.title("Which Platform NA like the most (Top 10)")
plt.xlabel("Platform")
plt.ylabel("NA_Sales")
plt.show()
```
